```python
#First, we import the necesary packages to read the files
import os
import pandas as pd
```

```python
def import_data_sbs_banks(start_month, start_year, end_month=None, end_year=None):  
    # Convert inputs to integers if they are provided as strings  
    start_month = int(start_month)  
    start_year = int(start_year)  
    
    # If end_month and end_year are not provided, set them to start_month and start_year  
    if end_month is None:  
        end_month = start_month  
    else:  
        end_month = int(end_month)  
        
    if end_year is None:  
        end_year = start_year  
    else:  
        end_year = int(end_year)  

    # Initialize a dictionary to store DataFrames  
    data_dict = {}  
    folder_path = '../../_data/sbs/B_RawData/bancos/'  
    
    # Iterate through the files in the specified folder  
    for file_name in os.listdir(folder_path):  
        # Check if the file starts with 'table_clean_' and ends with '.xlsx'  
        if file_name.startswith("table_clean_") and file_name.endswith('.xlsx'):  
            # Remove the '.xlsx' suffix from the file name and split it to get month and year  
            file_name = file_name.rsplit(".", 1)[0]  
            parts = file_name.split('_')  
            
            if len(parts) >= 3:  # Ensure we have enough parts  
                day, month, year = int(parts[-3]), int(parts[-2]), int(parts[-1])  # Extracting day, month, and year  

                # Check if the current year and month fall within the specified range  
                if (start_year < year or (start_year == year and start_month <= month)) and \
                   (year < end_year or (year == end_year and month <= end_month)):   
                    
                    file_path = os.path.join(folder_path, file_name + '.xlsx')  
                    df = pd.read_excel(file_path)  

                    # Add the DataFrame to the nested dictionary  
                    if year not in data_dict:  
                        data_dict[year] = {}  
                    data_dict[year][month] = df  

    # Sort the dictionary by year and month  
    sorted_data_dict = {year: dict(sorted(months.items())) for year, months in sorted(data_dict.items())}  

    return sorted_data_dict  
```
