```python
import os
import pandas as pd
```

Pasos para generar la función:

```python
# Valid years and months
valid_years = set(range(2001, 2023))
valid_months = set(range(1, 13))

# Directory for data
data_directory = "../../_data/sbs/B_RawData/bancos"
```

```python
def import_data_sbs_banks(start_month, start_year, end_month=None, end_year=None):
    """
    Load data from SBS bank files for a given month and year, or for a date range.

    :param start_month: Starting month (1-12) as a string.
    :param start_year: Starting year as a string.
    :param end_month: (Optional) Ending month as a string.
    :param end_year: (Optional) Ending year as a string.
    :return: A nested dictionary with years as keys, months as sub-keys, and DataFrames as values.
    """
    
    # Convert input parameters from strings to integers
    start_month = int(start_month)
    start_year = int(start_year)
    end_month = int(end_month) if end_month else None
    end_year = int(end_year) if end_year else None

   # Initialize dictionary for storing dataframes
    nested_dict = {}
    
    # Ensure the start month and year are valid
    if start_month not in valid_months or start_year not in valid_years:
        raise ValueError("Invalid start month or year.")
    if end_month and end_month not in valid_months:
        raise ValueError("Invalid end month.")
    if end_year and end_year not in valid_years:
        raise ValueError("Invalid end year.")
    
    # Set the end month and year to start values if they are not provided
    if end_month is None:
        end_month = start_month
    if end_year is None:
        end_year = start_year
    
    # Iterate through the range of years
    for year in range(start_year, end_year + 1):
        # Initialize the year in nested_dict if not present
        if year not in nested_dict:
            nested_dict[year] = {}
        
        # Set start and end months based on the year
        start = start_month if year == start_year else 1
        end = end_month if year == end_year else 12
        
        # Iterate through the range of months
        for month in range(start, end + 1):
            # Loop through days 1 to 31
            for day in range(1, 32):  
                # Construct the file name and its full path
                file_name = f'table_clean_{day}_{month}_{year}.xlsx'
                full_path = os.path.join(data_directory, file_name)
                
                # Convert backslashes to forward slashes
                formatted_path = os.path.normpath(full_path).replace("\\", "/")
                
                if os.path.isfile(formatted_path):
                    formatted_month = f'{month}'
                    if formatted_month not in nested_dict[year]:
                        nested_dict[year][formatted_month] = {}
                    nested_dict[year][formatted_month][f'{day}'] = pd.read_excel(formatted_path)
    
    return nested_dict
```
