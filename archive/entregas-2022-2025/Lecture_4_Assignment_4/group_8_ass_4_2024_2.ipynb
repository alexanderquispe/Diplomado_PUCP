```python
import os
import pandas as pd  

def import_data_sbs_banks(start_month, start_year, end_month=None, end_year=None):
    # end_month and end_year are not required by default
    if end_month is None:  # If an end date is not specified, the start date will be used
        end_month = start_month
    if end_year is None:
        end_year = start_year

    data = {}  # Dictionary to store dataframes

    # Convert to integers for easier comparisons
    start_month, start_year, end_month, end_year = map(int, [start_month, start_year, end_month, end_year])

    # Relative path to the folder that contains the files
    base_path = r"../../_data/sbs/B_RawData/bancos"

    # Browse the files in the folder
    for filename in os.listdir(base_path):
        # Check if the file has the expected format
        if filename.startswith('table_clean_') and filename.endswith('.xlsx'):
            # Extract day, month and year from file name
            parts = filename.replace('table_clean_', '').replace('.xlsx', '').split('_')
            if len(parts) == 3:
                file_day, file_month, file_year = map(int, parts)
                # Check if the file is in the specified date range
                
                file_date = (file_year, file_month)
                start_date = (start_year, start_month)
                end_date = (end_year, end_month)
                
                if start_date <= file_date <= end_date:
                    file_path = os.path.join(base_path, filename)
                    # Read the excel file
                    df = pd.read_excel(file_path)
                    file_path = os.path.join(base_path, filename)
                    df = pd.read_excel(file_path)
                    # Save in the nested dictionary
                    if file_year not in data:
                        data[file_year] = {}
                    data[file_year][file_month] = df
    return data

```
