```python
#7. If conditions with multiple expressions:

while True:
    try:
        t = float(input("Enter temperature in Celsius: "))
        break
    except ValueError:
        print("Please enter a valid number.")

if t < 0:
    print("It is freezing.")
elif 0 <= t <= 20:
    print("It is cold.")
elif 21 <= t <= 30:
    print("It is warm.")
else:
    print("It is hot.")
```

```python
#8. Logical operators

while True:
    try:
        gpa = float(input("GPA (e.g., 3.6): "))
        break
    except ValueError:
        print("Please enter a valid number for GPA.")

# Extracurricular
extracurricular = input('Extracurricular activities? (Yes/No): ').strip()

# Community service hours
while True:
    try:
        hours = int(input("Community service hours (integer): "))
        break
    except ValueError:
        print("Please enter a whole number for hours.")

if gpa > 3.5 and (extracurricular == "Yes" or hours > 50):
    print("Eligible for scholarship.")
else:
    print("Not eligible for scholarship.")
    
```

```python
#9. Python Identity Operators ---
list1 = [1, 2, 3]
list2 = [1, 2, 3]
list3 = list1

print("list1 is list2:", list1 is list2)   # False: objetos distintos en memoria
print("list1 is list3:", list1 is list3)   # True: misma referencia/objeto
print("list1 == list2:", list1 == list2)   # True: contenidos iguales

print("\nExplanation:")
print("- list1 is list2 → False: same values, but different memory locations (different objects).")
print("- list1 is list3 → True: list3 points to exactly the same object as list1.")
print("- list1 == list2 → True: their contents are equal, even if they are different objects.")
```

```python
#10. Nested If Statement:
while True:
    try:
        score = int(input("Enter student score (0-100): "))
        if 0 <= score <= 100:
            break
        else:
            print("Score must be between 0 and 100.")
    except ValueError:
        print("Please enter a whole number between 0 and 100.")

if score >= 90:
    print("A")
elif 80 <= score <= 89:
    if score == 85:
        print("B+")
    else:
        print("B")
elif 70 <= score <= 79:
    print("C")
else:
    print("Fail")
```

```python
# 11. For Loop in NumPy
import numpy as np

# NumPy array
arr = np.array([10, 20, 30, 40, 50])

# Iteration over each element and print its value multiplied by 2
print("11.1. Each value multiplied by 2:")
for num in arr:
    print(num * 2)

# New NumPy array instead
results = []
for num in arr:
    results.append(num * 2)
new_array = np.array(results)
print("11.2. New array with the results:", new_array)
```

```python
# 12. For Loop in List
words = ["python", "loop", "list", "iteration"]

# The length of each word
print("12.1. Length of each word:")
for word in words:
    print(len(word))

# Same loop using a list comprehension
lengths = [len(word) for word in words]
print("12.2. List of word lengths:", lengths)
```

```python
# 13. For Loop in Dictionary
student_scores = {"Alice": 85, "Bob": 92, "Charlie": 78, "Diana": 88}

# Each student's name and their score
print("13.1. Student names and scores:")
for name, score in student_scores.items():
    print(f"{name}: {score}")

# The names of students who scored above 80
print("13.2. Students who scored above 80:")
for name, score in student_scores.items():
    if score > 80:
        print(name)
```

```python
# 14. For Loop using range()
print("14.1. Even numbers between 1 and 20:")
for num in range(2, 21, 2):
    print(num)

# Sum of the even numbers while iterating
sum_even = 0
for num in range(2, 21, 2):
    print(num)
    sum_even += num
print("Sum of even numbers:", sum_even)
```

```python
# 15. Iterating over a Pandas DataFrame (ENAHO dataset)

import pandas as pd

# P41601: "How much was the total amount paid for the purchase or service?"

try:
    df400 = pd.read_csv("ENAHO01A-2023-400.csv", encoding="ISO-8859-10", low_memory=False)
    df400["P41601"] = pd.to_numeric(df400["P41601"], errors="coerce")
    
    print("Summary statistics for P41601:")
    print(df400["P41601"].describe())
    print("\nTop 10 largest values in P41601:")
    print(df400["P41601"].sort_values(ascending=False).head(10))
    
    print("\nCount of values > 5000 (including 99999.9):", (df400["P41601"] > 5000).sum())
    
    print("Count of values equal to 99999.9:", (df400["P41601"] == 99999.9).sum())
    
    mask = (df400["P41601"] > 5000) & (df400["P41601"] != 99999.9)
    print("Count of values > 5000 and not 99999.9:", mask.sum())
    
    filtered = df400.loc[mask, "P41601"]
    print("\nFiltered values (real amounts > 5000):")
    print(filtered if not filtered.empty else "No real values greater than 5000 found (excluding 99999.9).")
    
except Exception as e:
    print("Error:", e)

# For-loop version 
print("Values greater than 5000 in 'P41601' (excluding 99999.9), using for loop:")
for value in df400["P41601"]:
    if pd.notnull(value) and value > 5000 and value != 99999.9:
        print(value)

print("Values greater than 5000 in 'P41601' (excluding 99999.9), using boolean indexing:")
print(df400[(df400["P41601"] > 5000) & (df400["P41601"] != 99999.9)]["P41601"])
```
