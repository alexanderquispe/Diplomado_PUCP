###  ***Perform the following modifications:***
####  A. Change the data type of a variable
####  B. Modify some values in a specific column

```python
# A. Change the data type of a variable
```

```python
# Verificar tipo actual de la columna
print(hous_sub['UBIGEO'].dtype)
```

```python
# Conversión de variable numérica "int" a "float"
hous_sub['UBIGEO'] = pd.to_numeric(hous_sub['UBIGEO'], errors='coerce')
```

```python
# Verificar tipo actual
print("Tipo original de UBIGEO:", hous_sub['UBIGEO'].dtype)

# Conversión de int a float
hous_sub['UBIGEO'] = hous_sub['UBIGEO'].astype(float)

# Verificar tipo después de la conversión
print("Tipo después de la conversión:", hous_sub['UBIGEO'].dtype)

# Mostrar primeras filas para comprobar
print(hous_sub[['UBIGEO']].head())
```

```python
# Verificar tipo actual
print("Tipo original de UBIGEO:", hous_sub['UBIGEO'].dtype)

# Conversión de float a texto (string)
hous_sub['UBIGEO'] = hous_sub['UBIGEO'].astype(str)

# Verificar tipo después de la conversión
print("Tipo después de la conversión:", hous_sub['UBIGEO'].dtype)

# Mostrar primeras filas para comprobar
print(hous_sub[['UBIGEO']].head())
```

```python
# B. Modify some values in a specific column
```

#### Valores originales de variable ESTRATO según Diccionario ENAHO 2023
1. De 500 000 a más habitantes
2. De 100 000 a 499 999 habitantes
3. De 50 000 a 99 999 habitantes
4. De 20 000 a 49 999 habitantes
5. De 2 000 a 19 999 habitantes
6. De 500 a 1 999 habitantes
7. Área de Empadronamiento Rural (AER) Compuesto
8. Área de Empadronamiento Rural (AER) Simple

Rango : 1 – 8

```python
# Reemplazar valores numéricos por descripciones abreviadas
hous_sub['ESTRATO'] = hous_sub['ESTRATO'].replace({
    1: '>= 500k hab.',
    2: '100k - 499k hab.',
    3: '50k - 99k hab.',
    4: '20k - 49k hab.',
    5: '2k - 19k hab.',
    6: '500 - 1,999 hab.',
    7: 'AER Compuesto',
    8: 'AER Simple'})
```
